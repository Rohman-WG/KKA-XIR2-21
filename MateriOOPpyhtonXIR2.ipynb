Pentingnya Mempassing Objek Utuh: Agar method serang memiliki akses penuh ke seluruh atribut (seperti hp dan attack_power) sekaligus dapat mengeksekusi method internal milik lawan (seperti lawan.diserang()) secara presisi.

```python
# ========================================================
# LATIHAN 3: Pewarisan (Inheritance)
# ========================================================

class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# Class Mage adalah anak dari class Hero
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} (Mage) | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2) # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")

# -- Main Program --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)         # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond) # Skill khusus Mage
```

Tugas Analisis 3
1. Error: AttributeError: 'Mage' object has no attribute 'name'.  Mengapa terjadi: Tanpa super(), constructor milik class induk (Hero) tidak pernah dieksekusi. Akibatnya, atribut name, hp, dan attack_power tidak terdaftar pada objek Mage.  
2. Peran super(): Mengarahkan dan mengeksekusi constructor atau method milik class induk (Parent) agar child class dapat mewarisi dan menginisialisasi atribut induk secara otomatis tanpa redundansi kode.

```python
# ========================================================
# LATIHAN 4: Enkapsulasi (Mengamankan Data HP)
# ========================================================

class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0 # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
hero1 = Hero("Layla", 100)

hero1.set_hp(-50)
print(f"HP setelah diset -50 (melalui setter): {hero1.get_hp()}")

# --- Percobaan Hacking & Analisis ---
print("\n--- Uji Akses Langsung / Hacking ---")
try:
    print(hero1.__hp)
except AttributeError as e:
    print(f"Akses langsung hero1.__hp gagal! Error: {e}")

# Akses dengan Name Mangling (Hacking):
print(f"Mencoba akses paksa (Name Mangling): {hero1._Hero__hp}")
```

Tugas Analisis 4
1. Name Mangling: Hasil: Nilai HP tetap bisa diakses.  Penyebab: Python secara internal mengubah nama atribut __hp menjadi _Hero__hp.  Alasan Dilarang: Mengaksesnya secara langsung merusak prinsip Encapsulation dan dapat melewati logika keamanan/validasi data.  
2. Uji Validasi (Set HP Negatif): Hasil: Nilai HP Hero akan bernilai negatif (-100).  Pentingnya Setter: Method Setter bertindak sebagai filter validasi untuk mencegah masukan data yang tidak valid (misal: HP bernilai negatif atau melebihi batas maks) demi menjaga integritas data.

```python
# ========================================================
# LATIHAN 5: Abstraction & Interface (Membuat Kontrak/Standar)
# ========================================================

from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")

class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")

# Uji Coba
h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()

# Demonstrasi Error Instansiasi Abstract Class:
print("\n--- Percobaan Instansiasi GameUnit ---")
try:
    unit = GameUnit()
except TypeError as e:
    print(f"Gagal membuat objek GameUnit! Error: {e}")
```

Tugas Analisis 5
1. Melanggar Kontrak:Error: TypeError: Can't instantiate abstract class Hero with abstract method serang.  Arti & Konsekuensi: Class Hero belum memenuhi kontrak dari GameUnit. Konsekuensinya, class Hero dianggap masih abstrak dan program menolak pembuatan objek darinya.  
2. Instansiasi Abstract Class:Mengapa Dilarang: GameUnit adalah blueprint abstrak yang method-nya belum memiliki implementasi/isi nyata.  Kegunaan: Sebagai cetakan standar (interface) yang menjamin seluruh class turunan memiliki struktur method yang seragam.

```python
# ========================================================
# LATIHAN 6: Polymorphism (Fleksibilitas Interaksi)
# ========================================================

class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")

class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")

class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")

class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")

# --- Jawaban Analisis 6.1: Menambahkan Class Healer ---
class Healer(Hero):
    def serang(self):
        print(f"{self.nama} tidak menyerang, tapi menyembuhkan teman!")

# Penerapan Polymorphism
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord"),
    Healer("Estes")  # Penambahan class baru
]

print("--- PERANG DIMULAI ---")
for pahlawan in pasukan:
    pahlawan.serang()
```

Tugas Analisis 6
1. Uji Skalabilitas (Class Healer):Hasil: Program berjalan lancar tanpa perlu merubah kode looping utama.  Keuntungan Polimorfisme: Memudahkan pengembangan program (extensible); fitur/karakter baru dapat ditambahkan kapan saja tanpa merusak sistem pemanggilan kode yang sudah ada.  
2. Konsistensi Penamaan Method:Hasil: Objek Archer akan mengeksekusi method serang() bawaan class induk Hero (serangan kosong) alih-alih method tembak_panah().  Mengapa Harus Sama: Polimorfisme mengandalkan nama method yang identik agar loop pemanggil dapat menjalankan tindakan yang tepat secara otomatis tanpa perlu mengecek tipe objeknya satu per satu.

```python
# ========================================================
# TUGAS PROYEK INTEGRASI: Sistem Manajemen Kamar Hotel "MyEdotel"
# ========================================================

from abc import ABC, abstractmethod

# 1. Abstraction (Abstract Class)
class KamarHotel(ABC):
    def __init__(self, nama_kamar, harga_dasar, stok_awal):
        self.nama_kamar = nama_kamar
        # Encapsulation: Private Attribute
        self.__harga_dasar = harga_dasar
        self.__stok = 0
        self.set_stok(stok_awal) # Validasi via setter saat inisialisasi

    # Getter & Setter untuk Harga Dasar
    def get_harga_dasar(self):
        return self.__harga_dasar

    # Getter untuk Stok
    def get_stok(self):
        return self.__stok

    # Setter / Tambah Stok dengan Validasi
    def set_stok(self, jumlah):
        if jumlah < 0:
            print(f"Gagal update stok {self.nama_kamar}! Stok tidak boleh negatif ({jumlah}).")
        else:
            self.__stok = jumlah
            print(f"Berhasil menambahkan stok {self.nama_kamar}: {self.__stok} unit.")

    # Abstract Methods (Kontrak)
    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


# 2. Inheritance & Polymorphism: KamarDeluxe
class KamarDeluxe(KamarHotel):
    def __init__(self, nama_kamar, harga_dasar, stok_awal, fasilitas):
        super().__init__(nama_kamar, harga_dasar, stok_awal)
        self.fasilitas = fasilitas

    def hitung_pajak(self):
        return self.get_harga_dasar() * 0.10 # Pajak 10%

    def hitung_harga_total(self, jumlah_malam):
        pajak = self.hitung_pajak()
        return (self.get_harga_dasar() + pajak) * jumlah_malam

    def tampilkan_detail(self):
        pajak = self.hitung_pajak()
        print(f"[DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}")
        print(f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | Pajak (10%): Rp {pajak:,.0f}")


# 3. Inheritance & Polymorphism: KamarStandard
class KamarStandard(KamarHotel):
    def __init__(self, nama_kamar, harga_dasar, stok_awal, kapasitas):
        super().__init__(nama_kamar, harga_dasar, stok_awal)
        self.kapasitas = kapasitas

    def hitung_pajak(self):
        return self.get_harga_dasar() * 0.05 # Pajak 5%

    def hitung_harga_total(self, jumlah_malam):
        pajak = self.hitung_pajak()
        return (self.get_harga_dasar() + pajak) * jumlah_malam

    def tampilkan_detail(self):
        pajak = self.hitung_pajak()
        print(f"[STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}")
        print(f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | Pajak (5%): Rp {pajak:,.0f}")


# 4. Fungsi Fitur Pemesanan
def proses_transaksi(daftar_pesanan):
    print("\nSTRUK PEMESANAN")
    total_tagihan = 0

    for idx, (kamar, malam) in enumerate(daftar_pesanan, start=1):
        print(f"{idx}. ", end="")
        kamar.tampilkan_detail()
        subtotal = kamar.hitung_harga_total(malam)
        total_tagihan += subtotal
        print(f"Menginap: {malam} malam | Subtotal: Rp {subtotal:,.0f}\n")

    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}")


# ========================================================
# EKSEKUSI ALUR PROGRAM (USER STORY)
# ========================================================

print("SETUP DATA KAMAR")
# a) Admin membuat data kamar Deluxe & Standard
kamar_deluxe = KamarDeluxe("Kamar Deluxe Sea View", 1500000, 10, "Private Pool")

# b) Admin mencoba mengisi stok negatif (akan ditolak)
kamar_standard = KamarStandard("Kamar Standard Superior", 500000, -5, "2 Orang")
# Memperbaiki stok kamar standard
kamar_standard.set_stok(20)

# c) & d) Tamu memesan 2 malam Kamar Deluxe dan 1 malam Kamar Standard
pesanan_tamu = [
    (kamar_deluxe, 2),
    (kamar_standard, 1)
]

proses_transaksi(pesanan_tamu)
```
